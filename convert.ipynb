```python
import numpy as np
import json

# --- Load normalization stats ---
norm = np.load("normalization_stats.npz")
mean = norm["mean"]
std = norm["std"]

# Save as JSON
with open("normalization.json", "w") as f:
    json.dump({
        "mean": mean.tolist(),
        "std": std.tolist()
    }, f)

print("Normalization JSON saved!")

# --- Load labels ---
labels = np.load("label_map.npy", allow_pickle=True)
labels_list = labels.tolist()  # convert to regular Python list

with open("label.json", "w") as f:
    json.dump(labels_list, f)

print("Label JSON saved!")
```
